# import thư viện

```python
import os
import numpy as np
import pandas as pd
import pandas
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

# load data

```python
# Dữ liệu mẫu
sentences = [
    "Tôi thích học khoá Computer Science.",
    "Bạn có đam mê khoa học máy tính không?",
    "Huy thích lập trình giải quyết vấn đề.",
    "An muốn lập trình game bằng Python.",
    "Mỹ có nhiều dự án Scratch hay.",
    "Em thấy lập trình giống với trò chơi giải đố.",
    "Không có gì đã bằng code chạy lần đầu tiên đúng luôn",
    "Sở thích của mình là lập trình các dự án hay"
]
print(sentences)
```

# tokenization

```python
# tao tokenizer
tokenizer = Tokenizer(num_words=100, oov_token="<OOV>")

# huan luyen tokenizer voi du lieu
tokenizer.fit_on_texts(sentences)
```

```python
# in tu dien
from pprint import pprint
word_index = tokenizer.word_index
pprint(word_index)
```

```python
# su dung tu dien (tokenizer) da duoc huan luyen de tokenize 3 cau dau tien
sequences = tokenizer.texts_to_sequences(sentences)

# in ra ket qua tokenize 3 cau dau
for i in range(3):
    print(sentences[i])
    print(sequences[i])
    print()
```

# test

```python
test_data = [
    "Nam thích kem.",
    "An thích Computer Science"
]
```

```python
test_seq = tokenizer.texts_to_sequences(test_data)

for ind, s in enumerate(test_data):
    print(s) # in cau goc
    print(test_seq[ind]) # in token cua cau
    print()
```

```python
# Xu ly cau ve cung 1 do dai
from tensorflow.keras.preprocessing.sequence import pad_sequences

padded = pad_sequences(test_seq, maxlen=10, padding='post', truncating='post')
print(padded)
```
