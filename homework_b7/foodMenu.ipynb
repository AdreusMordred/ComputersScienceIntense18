# Setup

```python
import os
import pandas as pd
from dotenv import load_dotenv
from google import genai
from google.genai import types
```

```python
env_path = ""
```

```python
load_dotenv(env_path)
google_api_key = os.getenv("GEMINI_API_KEY")
genai_client = genai.Client(api_key=google_api_key)
```

# System instruction

```python
MODEL_VERSION = "gemini-2.5-flash"
SYS_INSTRUCTION = """
Bạn là trợ lý AI của một nhà hàng.

Nhiệm vụ:
- Tạo thực đơn từ danh sách món ăn được cung cấp.
- MỖI MÓN ĂN PHẢI NẰM TRÊN MỘT DÒNG RIÊNG.
- Chỉ được xuất theo đúng format:

<Tên món> - <Mô tả món>

Ví dụ:
Pizza New York - Pizza phong cách New York với đế bánh giòn.
Salad - Món salad tươi mát với rau củ và nước sốt.

Không viết lời mở đầu.
Không viết lời kết.
Không gộp nhiều món vào cùng một câu.
"""
```

# Danh sách món ăn

```python
menu = [
    "Pizza New York",
    "Salad",
    "Tomato Spaghetti",
    "Tiramisu"
]
```

# Prompt

```python
prompt = f"""
Hãy tạo menu cho nhà hàng từ các món sau:

{', '.join(menu)}

Yêu cầu:
Tên món - Mô tả món
"""
```

# Gọi Gemini

```python
answer = genai_client.models.generate_content(
    model=MODEL_VERSION,
    contents=prompt,
    config=types.GenerateContentConfig(
        system_instruction=SYS_INSTRUCTION,
        temperature=0.2,
        max_output_tokens=1000,
    ),
)
```

# Kết quả

```python
from IPython.display import Markdown

Markdown(answer.text)
```
